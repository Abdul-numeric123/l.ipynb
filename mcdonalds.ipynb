Import Libraries

```python
import pandas as pd
import numpy as np
```

 Load the Dataset

```python
df = pd.read_csv("McDonalds_Financial_Statements.csv")
```

Initial Exploration

```python
print(df.head())             # Preview first few rows
print(df.info())             # Check data types and non-null counts
print(df.describe())         # Summary statistics
print(df.columns)            # List column names
```

Clean Column Names


```python

# Clean column names: strip spaces, replace spaces with underscores, and remove $, %, ()
df.columns = (
    df.columns
    .str.strip()
    .str.replace(r'\s+', '_', regex=True)
    .str.replace(r'[()$%]', '', regex=True)
    .str.lower()
)

print("Column names cleaned:", df.columns.tolist())
```

Handle Missing Values

```python
# Drop columns with >50% missing values
df = df.dropna(thresh=len(df) * 0.5, axis=1)

# Forward fill remaining missing values
df = df.fillna(method='ffill')

print("Missing values handled.")
```

Convert String Values to Numeric

```python
# Convert all object (string) columns to numeric where possible
for col in df.columns:
    if df[col].dtype == 'object':
        df[col] = df[col].astype(str).str.replace(',', '').str.replace('$', '').str.replace('%', '')
        df[col] = pd.to_numeric(df[col], errors='coerce')

print("String numeric columns converted.")
```

Drop Duplicates

```python
df = df.drop_duplicates()
print("Duplicates removed.")
```

Save Cleaned Data

```python
# Save the cleaned dataset
output_path = r"C:\Users\DELL\Downloads\twitter.py\McDonalds_Financial_Statements_CLEANED.csv"
df.to_csv(output_path, index=False)

print("Cleaned file saved to:", output_path)
```
